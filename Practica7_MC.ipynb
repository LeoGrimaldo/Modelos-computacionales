**Sorting problem**

```python
def insertion_sort(A):
  for j in range(1, len(A)):
    pivote=A[j]
    #i es el valor anterior inmediato a j
    i=j-1
    while i>= 0 and A[i] > pivote: #mientras i<= 0 y i es mayor al pivote
      A[i+1] = A[i] #Actualiza el valor de i
      i = i-1
    A[i+1] = pivote
    print(A)
  return(A)
```

```python
#Ejemplo
lista_pro = [5,2,3,1,6,4]
insertion_sort(lista_pro)
#print(lista_pro)
```
