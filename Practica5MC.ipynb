Hacer una función que calcule la suma de todos los elemntos de una lista sin importar su tamaño. Una con ciclo for y otra con while.


```python
def suma_en_lista(lista):
    suma_total = 0          #Declaramos la variable que corresponderá al valor de la suma
    for numeros in lista:   #Consideramos a todos los elementos de la lista
      suma_total += numeros #Suma total va acumulando el sumar cada elemento de la lista
    return suma_total
```

## Recordatorio: modelo logístico

Suposiciones:\
$i)$ Una población de individuos se reproduce con una tasa de reproducción $r$\
$ii)$ Compiten por un recurso*\
$iii)$ Hay un número máximo de individuos posibles $k$, llamada *capacidad de carga*

La ecuación es de la forma:

$$N_{t+1} = rN_t - \frac{rN_t ^2}{k} = rN_t \left( 1- \frac{N_t}{k} \right)$$

Donde:

$N$ = cantidad de individuos.\
$r$ = tasa de reproducción.\
k = capacidad de carga\
$t$ = tiempo

```python
#Definimos nuestra función logística
# depende de 'r'(parámetro) y de 'x' 'variable'

def logistica(r,N):
    return r*N*(1-N)
```

```python
#Declaramos nuestra condición inicial como elemento de una lista
x1=[0.05]

#Declaramos nuestro parámetro "r"
r=2.75

#OBSERVACIÓN: Lo siguiente utiliza comprensión de listas

for i in range(30):                       #Iteramos la regla de correspondencia de la ecuación logística 30 veces
    x1.append(logistica(r,x1[-1]))        #Añadimos el valor de f(r,x anterior)

plt.plot(x1,'.')                      #Graficamos estos valores, el punto indica el estilo de graficación
```
