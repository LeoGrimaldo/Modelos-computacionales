Crear números aleatorios

```python
#Para esto necesitamos importar la librería "random"
import random
#random.randit 1,100 elige un número aleatorio del 1 al 100
```

```python
#Repaso
#Hacer una lista de 800 números aleatorios entre 1 y 6
Lista_aleatorio= list(random.randint(1,6) for x in range(0,800))
print(Lista_aleatorio)

#Contar cuantas veces aparece cada número en la lista anterior
Numeros= (Lista_aleatorio.count(1),Lista_aleatorio.count(2),Lista_aleatorio.count(3),Lista_aleatorio.count(4),Lista_aleatorio.count(5),Lista_aleatorio.count(6))
print(Numeros)
```

Conjuntos

```python
#set nos sirve para generar conjuntos (agrupar y ordenar)
numeros=[1,2,3,4,5,5,6,7,7,7,7]
print(numeros)
conjunto=set(numeros)
print(conjunto)
```

```python
#Hacemos 2 listas
a=[2,3,3,4,5,5,5,6,7,7,8,8,5,9,12,12,13]
b=[1,1,1,2,3,4,5,5,6,6,8,8,9,10,11,12,12]

#Agrupamos los datos de cada lista
a=set(a)
b=set(b)

#Unión (todos los elementos en "a" o en "b")
print(a|b)

#Intersección (todos los elementos en "a" y en "b")
print(a&b)

#Diferencia (elementos que no comparten)
print(a-b)
#Elementos que están en a pero no en b o b pero no en a
print(a.difference(b))
print(b.difference(a))
```

Ejercicio final

```python
#Matriz de 100x100 con números aleatorios del 0 al 10 y cuenta cuantos 0 tiene

Ejercicio= [[random.randint(0,10) for i in range(0,100)] for j in range(0,100)]
print(len(Ejercicio))

Cero=[Ejercicio.count(0) for i in range(len(Ejercicio)) for j in range(len(Ejercicio[i]))]
print(Cero)

lista= [x for fila in Ejercicio for x in fila]

Conteo=lista.count(0)
print(Conteo)
```
