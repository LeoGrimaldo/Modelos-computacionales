# Programación Orientada a Objetos - Jueves 23 de septiembre

La Programación Orientada a Objetos (POO) es un paradigma que organiza el código agrupando datos (atributos) y comportamientos (métodos) en entidades llamadas objetos, creados a partir de plantillas definidas como clases.


A continuación se muestra el ejemplo de unos perros

```python
"""
__init__ es un método inicializador especial (a menudo llamado constructor)
que se invoca automáticamente cuando se crea una nueva instancia de una clase.
Su propósito principal es inicializar los atributos del objeto y establecer
su estado inicial asignando valores a las variables de instancia mediante
el parámetro 'self'.
"""

"""
Haremos una función para definir perros con cinco características:
nombre, raza, tos (t), fiebre (f), si es violento (v).
El objetivo es construir posteriormente una función para ver si tienen
rabia o no.
"""


# Definimos la clase "Perro"
# en esta clase con __init__ declararemos los objetos
# También añadiremos funciones para verificar o modificar características de los objetos (perros)

class Perro:


#Definimos la función que contruirá los objetos (perros)
  def __init__(self, nombre, raza, t, f, v): #El perro tendrá 5 características
      self.nombre = nombre      #Declaramos a qué será igual el parámetro 'nombre'
      self.raza = raza          #Declaramos a qué será igual el parámetro 'raza'

      if(t==1 or t==0):         # Damos condiciones para el parámetro 't' que representa 'tos'
          self.tos = t          # 't' tomará valor de 0 o 1 si de de clara ese valor
      else:                     # En otro caso se le asignará 1
          self.tos=1

      if(f==1 or f==0):         #Repetimos lo hecho con 't', pero ahora para fiebre (f)
          self.fiebre = f
      else:
          self.fiebre=1

      if(v==1 or v==0):         #Repetimos análogamente para 'v'
          self.violento = v
      else:
          self.violento=1


  def cambiaNombre(self, nuevoNombre):    #Definimos una función para cambiar nombre
      self.nombre=nuevoNombre

  def getRaza(self):                      #Definimos una función para obtener la raza
      return self.raza

  def imprimePerro(self):               #Función que nos da el nobre y raza del perro
      return "El perro se llama " + self.nombre + " y es de raza " + self.raza


  #Si el perro tiene al menos dos parámetros de 't', 'f' o 'v', entonces el perro
  #está enfermo
  def revisaSalud(self):
      #print(self.nombre)
      res=""
      if(self.tos==0 and self.fiebre==0 and self.violento==0):
          res= "Está sano"
      elif(self.tos==0 and self.fiebre==0 and self.violento==1):
          res= "Está sano"
      elif(self.tos==0 and self.fiebre==1 and self.violento==0):
          res="Está sano"
      elif(self.tos==0 and self.fiebre==1 and self.violento==1):
          res="Está enfermo"
      elif(self.tos==1 and self.fiebre==0 and self.violento==0):
          res="Está sano"
      elif(self.tos==1 and self.fiebre==0 and self.violento==1):
          res="Está enfermo"
      elif(self.tos==1 and self.fiebre==1 and self.violento==0):
          res="Está enfermo"
      else:
          res="Está enfermo"
      return res



### Comparamos la raza de dos perros para saber si son la misma
  def mismaRaza(self, p):
      print(p.nombre)
      if(self.getRaza()==p.getRaza()):
          print("Son de la misma raza")
      else:
          print("Son de diferente raza")

```
