## El Código anterior tienen dos problema si el usuario/consumidor debe ingresar una lista ¿Cuáles son y cómo se pueden corregir?

```python
#Queremos que si se ingresa un valor decimal lo considere no válido
def puestito(pedido):
  if len(pedido) == 3 and all(isinstance(x, int) for x in pedido) and all(isinstance(x >= 0) for x in pedido):
    cuenta = 20*pedido[0] + 22*pedido[1] + 25*pedido[2]
    print("Se debe pagar $" + str(cuenta))
  else:
    print("solo hay 3 cosas, checate")
```

```python
cliente3 = [3.5, 3, 1, 3]
puestito(cliente3)
```

```python
cliente4 = [-12, 21, 2]
puestito(cliente4)
```

##Repaso rápido de *for* y *while*


Puedes regresar a tu lugar

```python
#Definimos una lista:
lista = [2,3,4,'Hola', 4.5, False, 5, "granjear", "Aura", 4.63]
print(lista)
```

```python
### Recorremos con un ciclo for:

for elemento in lista:    #nota sobre los 'in'
    print(elemento)
```

```python
### Recorremos con while


i=0                     #Con while siempre necesitamos un contador
while(i<len(lista)):    #La condición es que 'i' sea menor a la longitud de la lista
    print(lista[i])     #Imprimimos el elemento 'i-ésimo'
    i+=1                #Se debe cerrar el loop while actualizando el valor
                        #del contador
```

#2. Algoritmos

Dos de los algoritmos más utilizados para ordenar datos en orden alfabético o por orden creciente/decreciente son: ***insertion sort*** y ***merge sort***.

\
El algoritmo ***insertion sort*** parte de una lista que se va recorriendo de izquierda a derecha que se deses ordenar sus valores de menor a mayor. Cada vez que se encuentre un número menor, este se recorre a la posición que le pertenece, es decir después del primer número (de izquierda a derecha) menor a él y antes del primer número mayor a él. Una vez hecho esto se repite hasta llegar al final de la lista. Este algoritmo suele usarlo la gente al jugar con cartas para acomodar su mano en orden creciente.

## Coming soon: modelo logístico

Suposiciones:\
$i)$ Una población de individuos se reproduce con una tasa de reproducción $r$\
$ii)$ Compiten por un recurso*
$iii)$ Hay un número máximo de individuos posibles $k$, llamada *capacidad de carga*

La ecuación es de la forma:

$$N_{t+1} = rN_t - \frac{rN_t ^2}{k} = rN_t \left( 1- \frac{N_t}{k} \right)$$

Donde:

$N$ = cantidad de individuos.\
$r$ = tasa de reproducción.\
k = capacidad de carga\
$t$ = tiempo
