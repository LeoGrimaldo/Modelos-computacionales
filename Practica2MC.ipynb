Hoy vamos a ver listas

```python
ligamx = ["Monarcas", "Mazatlán", "Jaguares", "Veracruz", "Leones_negros", "BUAP", "Toros_Neza"] #creamos una lista
print(ligamx[0])

print(len(ligamx)) #nos dice el número de elementos de nuestra lista

new_list = list(range(0,len(ligamx),1)) #Crea una lista con el numero de elementos de nuestra lista anterior
print(new_list)

#Al ser mutables, podemos modificar las listas
ligamx[-1] = "Mineros"
print(ligamx)

#Tambien podemos añadir elementos
ligamx.append("Toros_Neza")
print(ligamx)

#Podemos eliminar y traspasar elementos
desaparecidos = ligamx.pop()
print(ligamx)
print(desaparecidos)

#Transformar en mayúsculas
ligamx = [equipo.upper() for equipo in ligamx]
print(ligamx)
```

```python
#Una lista con los números del 0 al 20 solo con los pares
pares = list(range(0,21,2))
print(pares)
```

```python
#Tomar solo los pares
completa = [2,3,8,6,4,18,17,25,90,81]
pares = list(x for x in completa if x%2==0)
print(pares)

#Si queremos solo impares
impares = list(x for x in completa if x%2!=0)
print(impares)
```

```python
#Remplazar 0 por None
absorb = [0.1,0,0.3,0,0.35,0]
absorb_none=[item if item !=0 else None for item in absorb]
print(absorb)
print(absorb_none)
```
