Operaciones lógicas

```python
# Es como hacerle una pregunta y que sólo te puedan responder verdero/falso

y = 2
z = 9
print(y == z) # Preguntamos igualdad (NO CONFUNDIR CON =, nótese que son 2)
print(y < z) # Preguntamos si y es menor que z
print(y > z) # Preguntamos si y es mayor que z
print(y != z) # Preguntamos si y es diferente de z
print(y <= z) # Preguntamos si y es menor o igual que z
print(y >= z) # Preguntamos si y es mayor o igual que z
```

Operaciones con control de flujo

```python
# Operadores condicionales if, elif, else

perritos_en_casa = 0

if perritos_en_casa == 0: # Noten cómo después de la condición va un : y justo abajo la consecuencia
  print("Tristeza absoluta")
elif perritos_en_casa == 1: # Permite agregar una condición más
    print("Alguien me espera")
elif perritos_en_casa == 2:
    print("Casita llena de amor")
else: # El resultado si la condición no se cumple
    print("No me alcanza el dinero :(")

# Para los condicionales la indexación es súper importante, marca la secuencia lógica del flujo
## USTEDES háganlo con el número de perritos en casa ##
```

```python
# Bucles; uno de los más importantes, permite automatizar procesos recursivos
counter = 1
alumnxs = ["Pedro", "Ana", "Joel", "Yandel"]
for students in alumnxs: #para cada elemento en alumnos
  if students == "Joel":
    print(counter)
  else:
    counter += 1

#Nos dice la posición de Joel en la lista
```

```python
# Un ejemplo más

for i in range(0,10,1): # 10 elementos empezando por 0 con pasos de tamaño 1
  if i<=5:
    print (f"{i} like a snake!")
  else:
    print (f"{i} is to close to heaven") # f antes del string para poder colocar un objeto (número)

# Si es <= a 5 imprime "like a snake", si es >5 imprime "is to close to heaven"
```
